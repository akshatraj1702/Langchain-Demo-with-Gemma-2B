# LangChain Demo with Gemma 2B

This notebook runs the same LangChain + Ollama workflow as the Streamlit app, but directly inside Jupyter.

```python
import os

from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import OllamaLLM
```

```python
# Pulls keys/settings from .env
load_dotenv()

# LangSmith tracing is optional
if os.getenv("LANGCHAIN_API_KEY"):
    os.environ["LANGCHAIN_TRACING_V2"] = "true"
    os.environ["LANGCHAIN_PROJECT"] = os.getenv(
        "LANGCHAIN_PROJECT", "langchain-notes"
    )
```

```python
# Prompt template
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant. Please respond to the question asked"),
        ("user", "Question: {question}"),
    ]
)

prompt
```

```python
# Runs locally through Ollama, so there is no API cost for the model call
llm = OllamaLLM(model="gemma:2b")

# Converts the model's output into plain text
output_parser = StrOutputParser()
```

```python
# LCEL: output of one step becomes input to the next
chain = prompt | llm | output_parser

chain
```

```python
# Enter your question directly in Jupyter
input_text = input("What question do you have in mind? ")

if input_text:
    response = chain.invoke({"question": input_text})
    print(response)
```
